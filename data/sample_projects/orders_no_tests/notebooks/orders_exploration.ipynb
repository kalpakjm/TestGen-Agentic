# Orders exploration
Quick look at order statuses. Reusable logic lives in `src/orders_etl`; this notebook keeps one helper for ad-hoc analysis.

```python
from pyspark.sql import functions as F


def status_counts(df):
    """Number of orders per status, most frequent first."""
    return df.groupBy("status").count().orderBy(F.col("count").desc())
```

```python
%sql
SELECT 1
```

```python
orders = spark.read.csv("../../../sample_data/orders.csv", header=True)
status_counts(orders).show()
```
